Fix the errors

Keep only the countries from the list.

Cause we have an integer in one of the cells

```python
VALID_COUNTRIES = ['France', 'Côte d\'ivoire', 
                   'Madagascar', 'Bénin', 
                   'Allemagne', 'USA']
mask = ~data['pays'].isin(VALID_COUNTRIES)
data.loc[mask, 'pays'] = np.nan
```
